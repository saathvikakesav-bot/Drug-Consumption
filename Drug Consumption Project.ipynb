```python
def cap_outliers_iqr(dataframe, exclude_columns=None):
    dataframe = dataframe.copy()

    if exclude_columns is None:
        exclude_columns = []

    numeric_columns = dataframe.select_dtypes(include="number").columns

    for column in numeric_columns:
        if column in exclude_columns:
            continue

        q1 = dataframe[column].quantile(0.25)
        q3 = dataframe[column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        dataframe[column] = dataframe[column].clip(
            lower=lower_bound,
            upper=upper_bound
        )

    return dataframe


df = cap_outliers_iqr(
    df,
   exclude_columns=["ID", "default payment next month"]
)

print("Outlier capping completed.")
```

```python
def group_alcohol(value):
    level = int(value.replace("CL", ""))

    if level <= 2:
        return "Low"
    elif level <= 4:
        return "Moderate"
    else:
        return "High"

y_grouped = df["Alcohol"].apply(group_alcohol)

print(y_grouped.value_counts())
```

```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report
from sklearn.model_selection import train_test_split

X_grouped = X.copy()

X_train_g, X_test_g, y_train_g, y_test_g = train_test_split(
    X_grouped,
    y_grouped,
    test_size=0.20,
    random_state=42,
    stratify=y_grouped
)

group_model = ExtraTreesClassifier(
    n_estimators=500,
    max_depth=10,
    min_samples_split=5,
    class_weight='balanced',
    random_state=42
)

group_model.fit(X_train_g, y_train_g)

group_pred = group_model.predict(X_test_g)

print("Accuracy:",
      round(accuracy_score(y_test_g, group_pred) * 100, 2), "%")

print("Balanced Accuracy:",
      round(balanced_accuracy_score(y_test_g, group_pred) * 100, 2), "%")

print(classification_report(y_test_g, group_pred))
```
